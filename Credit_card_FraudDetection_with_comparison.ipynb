Data preprocessing

```python
X=df.drop("Class",axis=1)
y=df["Class"]
```

```python
X.head()
```

```python
df["Class"].isna().sum()
```

```python
df=df.dropna(subset=["Class"])
```

```python
X=df.drop("Class",axis=1)
y=df["Class"]
```

```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(X,y,random_state=42,stratify=y)
```

Featue Scaling

Standardization Scaling

```python
from sklearn.preprocessing import StandardScaler
scaler=StandardScaler()
x_train_scaled=scaler.fit_transform(x_train)
x_test_scaled=scaler.transform(x_test)
```

```python
x_train_scaled[:5]
```

Imbalance

```python
from imblearn.under_sampling import RandomUnderSampler
rus=RandomUnderSampler(random_state=42)
x_train_rus,y_train_rus=rus.fit_resample(x_train_scaled,y_train)
```

```python
y_train_rus.value_counts()
```

Model Train

```python
from sklearn.linear_model import LogisticRegression
model = LogisticRegression(max_iter=1000)  #
model.fit(x_train_rus, y_train_rus)
```

```python
y_pred=model.predict(x_test_scaled)
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,y_pred)
cm
```

```python
from sklearn.metrics import accuracy_score,precision_score,recall_score,f1_score
print("Accuracy Score",accuracy_score(y_test,y_pred))
print("Precision Score",precision_score(y_test,y_pred))
print("Recall Score",recall_score(y_test,y_pred))
print("F1 Score",f1_score(y_test,y_pred))
```

```python
from sklearn.metrics import classification_report
print(classification_report(y_test,y_pred))
```

```python
from sklearn.metrics import roc_auc_score
y_prob=model.predict_proba(x_test_scaled)[:,1]
roc_auc_score(y_test,y_prob)
#
```

Version-2


```python
from imblearn.under_sampling import RandomUnderSampler
rus=RandomUnderSampler(random_state=42,sampling_strategy=0.1)
x_train_v2,y_train_v2=rus.fit_resample(x_train_scaled,y_train)
#
```

```python
y_train_v2.value_counts()
```

```python
from sklearn.ensemble import RandomForestClassifier
model_v2=RandomForestClassifier(random_state=42)
model_v2.fit(x_train_v2,y_train_v2)
```

```python
y_pred_v2=model_v2.predict(x_test_scaled)
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(y_test,y_pred_v2)
cm
```

```python
print(classification_report(y_test,y_pred_v2))
print("Accuracy :",accuracy_score(y_test,y_pred_v2))
print("Precision :",precision_score(y_test,y_pred_v2))
print("Recall :",recall_score(y_test,y_pred_v2))
print("F1 Score :",f1_score(y_test,y_pred_v2))
```

## Model Comparison

```python
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score
import matplotlib.pyplot as plt

# Comparing both models
models = ["Logistic Regression", "Random Forest"]

accuracy = [accuracy_score(y_test, y_pred), accuracy_score(y_test, y_pred_v2)]
precision = [precision_score(y_test, y_pred), precision_score(y_test, y_pred_v2)]
recall = [recall_score(y_test, y_pred), recall_score(y_test, y_pred_v2)]
f1 = [f1_score(y_test, y_pred), f1_score(y_test, y_pred_v2)]

results = pd.DataFrame({
    "Accuracy": accuracy,
    "Precision": precision,
    "Recall": recall,
    "F1 Score": f1
}, index=models)

results
```

```python
results.plot(kind="bar", figsize=(10, 5))
plt.title("Logistic Regression vs Random Forest")
plt.ylabel("Score")
plt.ylim(0, 1)
plt.xticks(rotation=0)
plt.legend(loc="lower right")
plt.tight_layout()
plt.show()
```
